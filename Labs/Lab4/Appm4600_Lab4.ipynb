```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Subroutines

def bisection(f,f1,f2,a,b,Nmax):
     # first verify there is a root we can find in the interval
    fa = f(a)
    fb = f(b);
    if (fa*fb>0):
        ier = 1.0
        astar = a
        return [astar, ier]
        
    # verify end points are not a root
    if (fa == 0):
        astar = a
        ier = 0
        return [astar, ier]
    
    if (fb ==0):
        astar = b
        ier = 0
        return [astar, ier]

    d = 0.5*(a+b)
    fd = f(d)
    f1d = f1(d)
    f2d = f2(d)
    g1d = (f2d*fd)/(f1d**2)

    count = 0
    while (abs(g1d) >= 1):
        if count > Nmax:
            astar = d
            ier = 1.1
            return [astar, ier]
            
        if (fd ==0):
            astar = d
            ier = 0
            return [astar, ier]
        
        if (fa*fd<0):
            b = d
        else:
            a = d
            fa = fd
            d = 0.5*(a+b)
        
        fd = f(d)
        f1d = f1(d)
        f2d = f2(d)
        g1d = (f2d*fd)/(f1d**2)
        count = count + 1

    astar = d
    ier = 0
    print('count = ', count)
    return [astar, ier]

def newton(f,fp,p0,tol,Nmax):
    p = np.zeros(Nmax+1);
    p[0] = p0
    
    for it in range(Nmax):
        p1 = p0-f(p0)/fp(p0)
        p[it+1] = p1
        if (abs(p1-p0) < tol):
            pstar = p1
            info = 0
            return [p,pstar,info,it]
        p0 = p1

    pstar = p1
    info = 1.2
    return [p,pstar,info,it]
```

```python
# Driver

def driver():
    Nmax = 100
    tol = 1.e-14
    
    f = lambda x: np.exp(x**2 + 7*x - 30) - 1
    f1 = lambda x: (2*x + 7) * (f(x) + 1)
    f2 = lambda x: (4*x**2 + 28*x + 51) * (f(x) + 1)
    a = 2
    b = 4.5
    p0, bisectInfo = bisection(f,f1,f2,a,b,Nmax)

    if (int(bisectInfo) == 1):
        return bisectInfo
    
    (p,pstar, NewtonInfo,it) = newton(f,f1,p0,tol,Nmax)
    
    print('the approximate root is', '%16.16e' % pstar)
    print('the error message reads:', '%d' % NewtonInfo)
    print('number of iterations:', '%d' % it)
```

```python
driver()
```
