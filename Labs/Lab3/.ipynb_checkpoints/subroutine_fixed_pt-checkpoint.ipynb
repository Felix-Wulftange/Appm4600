```python
# define routines
def fixedpt(f,x0,tol,Nmax):
   x = np.zeros((Nmax,1))
   ''' x0 = initial guess''' 
   ''' Nmax = max number of iterations'''
   ''' tol = stopping tolerance'''

   count = 0
   while (count < Nmax):
      x1 = f(x0)
      x[count,0] = f(x0)
      count = count +1
      if (abs(x1-x0) <tol):
         xstar = x1
         ier = 0
         return [xstar,ier,x]
      x0 = x1

   xstar = x1
   ier = 1
   return [xstar, ier,x]
   
```

```python
# Aitkens subroutine

def aitkens(p, tol, Nmax):
    pa = np.zeros((Nmax, 1))
    count = 0
    while (count < Nmax-2):
        pn = p[count]
        pn1 = p[count+1]
        pn2 = p[count+2]
        
        pan =  pn - (pn1 - pn)**2 / (pn2 - 2*pn1 + pn)
        pa[count, 0] = pan
        count += 1
        if (abs(pan - pa[count]) < tol):
            xstar = pan
            ier = 0
            return pa

        xstar = pan
        ier = 1
        return pa

aitkens(x1, 1e-6, 100)
```
